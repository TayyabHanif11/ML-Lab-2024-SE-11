## Task 6: List vs Tuple

Following are the three important differences:

1. Main difference is that the Lists are mutable (they can be changed after creation), while tuples are immutable (they cannot be changed after creation).
2. Lists use square brackets [], while tuples use parentheses ().
3. Lists are slower to create while tuples can be created/iterated fastly.

```python

# Demonstration of their difference, tuple v list

my_list = [10, 20, 30]
my_tuple = (10, 20, 30)

my_list[0] = 100

print("Q6 - Modified list:", my_list)
print("Q6 - Tuple:", my_tuple)
```

## Task 7: Extracting an email domain

```python
def get_domain(email):
    return email.split("@")[1]

email = "user@domain.com"

print("Q7:", get_domain(email))
```

## Task 8: Checking whether a string contains "dog"

```python
def find_dog(text):
    return "dog" in text.lower()

print("Q8:", find_dog("Is there a DOG here?"))
print("Q8:", find_dog("There is a cat here."))
```

## Task 9: Count occurrences of "dog"

```python
def count_dog(text):
    return text.lower().count("dog")

sentence = "This dog runs faster than that DOG. Another dog is sleeping."

print("Q9:", count_dog(sentence))
```

## Task 10: Speeding ticket function

```python
# On birthday, the driver receives an extra 5 mph allowance/relief.

def caught_speeding(speed, is_birthday):

    if is_birthday:
        speed = speed - 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10 Test 1:", caught_speeding(60, False))
print("Q10 Test 2:", caught_speeding(75, False))
print("Q10 Test 3:", caught_speeding(85, False))
print("Q10 Test 4:", caught_speeding(65, True))
print("Q10 Test 5:", caught_speeding(85, True))
print("Q10 Test 6:", caught_speeding(86, True))
```
